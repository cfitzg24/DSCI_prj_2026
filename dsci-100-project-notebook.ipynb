#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)

- Comment your code throughout

```r

tidy_NCAA_data <- NCAA_data |>
        filter(year >= 1984) |> #Filtered for after 1983 as data before 1983 has excessive N/A values
        select(-conference, - first_game_at_home, - division) |> #removed non-numerical variables containing unnecessary data.
        select(-tourney_losses, - tourney_finish, - total_wins, - total_losses, - total_wins_pct) |> #removed variables with data from after/during the conference -> cannot use tournament performance to predict tournament performance
        drop_na(conf_rank) |> #drops schools with no conference data => analyze only schools that have conference data
        mutate(season_wins = conf_wins + reg_wins)|>
        mutate(season_losses = conf_losses + reg_losses)|>
        mutate(season_wins_pct = season_wins *100 / (season_losses + season_wins)) #recalculated season totals without tournament data.

head(tidy_NCAA_data, 6) #shows first 6 rows of data
```

The method I plan to use to answer my research question is a K-nn regression model. This model choice comes from two factors, the desire to look at tournament progress as non-categorical and the need to reduce outliers. The desire to look at tournament progress as a numerical rather than a categorical variable leads to a regression based model that predicts a value rather than a category. The need to avoid outliers, such as those shown in visualization three where teams with abnormally low season win percentages progress to far rounds, drives the model to a K-nn based engine that allows prediction based on similar teams rather than including all teams in the prediction. Potential limitations include a lower number of datapoints in the higher rounds of the tournament as the tournament slowly eliminates team, potentially skewing the data downwards with our K-nn model. The processing of this data will begin with a 75% - 25% split between testing and training data, then a 5 fold cross-validation will be performed to determine the best K to use for the data. #### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.*

```r
# cutting uncessary data and season data that would double count
model_NCAA_data <- tidy_NCAA_data |>
        select(-school, - bid, - season_wins, season_wins_pct, season_losses)
#Split Data into testing & training sets
NCAA_split <- initial_split(model_NCAA_data, prop = .75, strata = tourney_wins)
NCAA_training <- training(NCAA_split)
NCAA_testing <- testing(NCAA_split)   

#Tuning K-Values Model
NCAA_tune <- nearest_neighbor(weight_func = "rectangular", neighbors = tune())|>
        set_engine("kknn")|>
        set_mode("regression")

#General K-nnRecipe
NCAA_recipe <- recipe(tourney_wins ~ ., data = NCAA_training)|>
        step_scale(all_predictors())|>
        step_center(all_predictors())
#5-fold cross-validation
NCAA_vfold <- vfold_cv(NCAA_training, v = 5, strata = tourney_wins)

#Determining best K value
k_vals <- tibble(neighbors = seq(from = 5, to = 250, by = 5))
NCAA_k <- workflow()|>
        add_recipe(NCAA_recipe)|>
        add_model(NCAA_tune)|>
        tune_grid(resamples = NCAA_vfold, grid = k_vals)|>
        collect_metrics() |>
        filter(.metric == "rmse")|>
        arrange(mean) |>
        slice(1)|>
        pull(neighbors)
NCAA_k




```

```r
#Model Specs
NCAA_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = NCAA_k)|>
    set_engine("kknn")|>
    set_mode("regression")

#Running the model
NCAA_fit <- workflow()|>
    add_recipe(NCAA_recipe)|>
    add_model(NCAA_spec)|>
    fit(data = NCAA_training)

#Validating on testing set
NCAA_summary <- NCAA_fit|>
        predict(NCAA_training)|>
        bind_cols(NCAA_training)
NCAA_metrics <- NCAA_summary |>
        metrics(truth = tourney_wins, estimate = .pred)
NCAA_metrics
```
